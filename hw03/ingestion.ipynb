```python
import duckdb

# Connect to an in-memory DuckDB instance
con = duckdb.connect()

# Install and load the Postgres extension (only needed the first time)
con.execute("INSTALL postgres;")
con.execute("LOAD postgres;")

# Attach your target PostgreSQL database
con.execute("ATTACH 'postgresql://root:root@localhost:5432/ny_taxi' AS pg_db (TYPE postgres);")

# Stream the Parquet file directly into the Postgres table
con.execute("""
    create table pg_db.yellow_taxi AS 
    SELECT * FROM read_parquet('./data/yellow_tripdata_2024-01.parquet');
""")
```

```python
import duckdb

# Connect to an in-memory DuckDB instance
con = duckdb.connect()

# Install and load the Postgres extension (only needed the first time)
con.execute("INSTALL postgres;")
con.execute("LOAD postgres;")

# Attach your target PostgreSQL database
con.execute("ATTACH 'postgresql://root:root@localhost:5432/ny_taxi' AS pg_db (TYPE postgres);")

# Stream the Parquet file directly into the Postgres table
for i in range(2,7):
    con.execute(f"""
        INSERT INTO pg_db.yellow_taxi 
        SELECT * FROM read_parquet('./data/yellow_tripdata_2024-0{i}.parquet');
    """)
```
